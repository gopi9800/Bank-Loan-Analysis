Analysis

Total Loan Applications

```python
total_loan_applications = df['id'].count()
print("Total Loan Applications:",total_loan_applications)
```

MTD Total Loan Applications

```python
latest_issue_date = df['issue_date'].max()
latest_issue_year = latest_issue_date.year
latest_issue_month = latest_issue_date.month
mtd_data = df[(df['issue_date'].dt.year == latest_issue_year) & (df['issue_date'].dt.month == latest_issue_month)]
mtd_loan_applications = mtd_data['id'].count()
print(f"Total Loan Applications for {latest_issue_date.strftime('%B%Y')}:",mtd_loan_applications)
```

Total Funded Amount

```python
total_funded_amount = df['loan_amount'].sum()
print("Total Funded Amount:",round(total_funded_amount/1000000,2),"millions")
```

MTD Total Funded Amount

```python
latest_issue_date = df['issue_date'].max()
latest_issue_year = latest_issue_date.year
latest_issue_month = latest_issue_date.month
mtd_data = df[(df['issue_date'].dt.year == latest_issue_year) & (df['issue_date'].dt.month == latest_issue_month)]
mtd_funded_amount = mtd_data['loan_amount'].sum()
print(f"Total Funded Amount for {latest_issue_date.strftime("%B %Y")}:",round((mtd_funded_amount/1000000),2),"millions")
```

Total Amount Received

```python
total_amount = df['total_payment'].sum()
print("Total Amount Funded:",round(total_amount/1000000,2),"millions")
```

MTD Total Amount Received

```python
latest_issue_date = df['issue_date'].max()
latest_issue_year = latest_issue_date.year
latest_issue_month = latest_issue_date.month
mtd_data = df[(df['issue_date'].dt.year == latest_issue_year) & (df['issue_date'].dt.month == latest_issue_month)]
mtd_amount = mtd_data['total_payment'].sum()
print(f"Total Funded Amount for {latest_issue_date.strftime("%B %Y")}:",round((mtd_amount/1000000),2),"millions")
```

Average Interest Rate

```python
avg_interest_rate = df['int_rate'].mean()
print("Average Interest Rate:",round(avg_interest_rate,2),"%")
```

Average Debt to Income Ratio(DTI)

```python
avg_dti_rate = df['dti'].mean() * 100
print("Average DTI Rate:",round(avg_dti_rate,2),"%")
```

Good Loan Metrics

```python
good_loans = df[df['loan_status'].isin(['Fully Paid','Current'])]
total_loan_applications = df['id'].count()
good_loan_applications = good_loans['id'].count()
good_loan_funded_amount = good_loans['loan_amount'].sum()
good_loan_received_amount = good_loans['total_payment'].sum()
good_loan_percentage = (good_loan_applications / total_loan_applications) * 100

print("Good Loan Applications:",good_loan_applications)
print("Good Loan Funded Amount:",round(good_loan_funded_amount/1000000,2),"millions")
print("Good Loan Funded Amount:",round(good_loan_received_amount/1000000,2),"millions")
print("Percentage of Good Loan Applications:",round(good_loan_percentage,2),"%")
```

Bad Loan Metrics

```python
bad_loans = df[df['loan_status'].isin(['Charged Off'])]
total_loan_applications = df['id'].count()
bad_loan_applications = bad_loans['id'].count()
bad_loan_funded_amount = bad_loans['loan_amount'].sum()
bad_loan_received_amount = bad_loans['total_payment'].sum()
bad_loan_percentage = (bad_loan_applications / total_loan_applications) * 100

print("Bad Loan Applications:",bad_loan_applications)
print("Bad Loan Funded Amount:",round(bad_loan_funded_amount/1000000,2),"millions")
print("Bad Loan Funded Amount:",round(bad_loan_received_amount/1000000,2),"millions")
print("Percentage of Bad Loan Applications:",round(bad_loan_percentage,2),"%")
```
